```python
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

mpl.rcParams['font.family'] = 'Arial'
mpl.rcParams['font.size'] = 10
```

```python
data_path = '../data/power/power.csv'
data = pd.read_csv(data_path)
data.head()
```

```python
timestamps = pd.to_datetime(data['Time'])
timestamps_seconds = (timestamps - timestamps.iloc[0]).dt.total_seconds()
power = data['Power (W)']
voltage = data['Voltage (V)']
current = data['Current (A)']

plt.figure(figsize=(8, 3))
plt.plot(timestamps_seconds, power, label='Power Consumption (W)', color='blue')
plt.xlabel('Time (s)')
plt.ylabel('Power (W)')

plt.xlim(0, 90)
plt.ylim(0.3, 0.4)

plt.tick_params(direction='in')

plt.savefig('../figures/SI/figs1-circuit_and_power/power_consumption_plot.png', dpi=600, bbox_inches='tight', transparent=True)
plt.show()
```
